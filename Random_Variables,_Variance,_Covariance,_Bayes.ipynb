- Sharding refers to the distribution of a logical database among different servers, known as shards, in such a way that a single server is never required to store or process the entire dataset. Each shard contains a part of the dataset, and all shards function together as a logical database.

This differs from replication, whereby the same data is stored in more than one server in case there are backups. In sharding, data is split; it focuses on capacity rather than replication, although, in reality, as will be explained later, sharding is always accompanied by replication

```python
#What is replication and how does it ensure high availability?
```

- Replication involves keeping several copies of the same data on different servers in such a way that in case one server crashes, there will be another that will continue the operation without loss of data or delay. Replication differs from sharding in the sense that sharding involves spreading out the database among servers while replication involves copying all data into every member.

How Replication Works in MongoDB: Replica Sets

Replica set refers to a number of MongoDB servers (members) that store the same data:

One Primary Member: The only member that writes data.
Secondary Members: They continuously copy data from the primary member and perform read operations (configured for that purpose).
Usually implemented using an odd number of members

-High variances indicate that customer spend varies greatly relative to the mean – there is a wide spread of values between those who spend nothing and those who spend much, with no concentration on the typical value.

For a business, what this implies is:

Average becomes an unreliable measure. If variances in spend are high, then claiming that "the average spend is $150" becomes deceptive – since the range of the actual spend might vary anywhere between $10 to $2,000, with no customer fitting into the average at all.
Indicates different customer segments.
```python
#Difference between normal distribution and skewed distribution
```

- Normal distribution

A symmetrical, bell-shaped curve; the mean = median = mode = center of the distribution.
Values are just as likely to fall above the mean as below by a certain value.
Empirical Rule: 68% of the data lies within 1 standard deviation of the mean, 95% within 2, and 99.7% within 3.

Skewed distribution

An asymmetrical distribution; where the tail on one side is longer or heavier than the other.
Mean, median, and mode do not coincide.
Right-skewed (positively skewed): a longer tail on the right. Values are concentrated on the left side of the distribution, but some larger values pull the mean above the median. Example: Income distribution, spending on customers.
Left-skewed (negatively skewed): a longer tail on the left. Values are concentrated on the right side of the distribution, but some extremely small values pull the mean below the median. Example: Retirement age distribution, test score on an easy exam.

```python
#Why do business metrics often show right-skewed distributions?
```

- The presence of a natural lower bound of zero and lack of an upper bound. Some variables, such as spending, revenue, duration of a session, and size of an order cannot be less than zero, but they have theoretically infinite maximum possible values. The asymmetry between the two bounds alone contributes to the right-skewed distribution as there is more space on the right side.
Small number of outliers with high values creates a tail. In most businesses, it happens that only a small number of customers, transactions, and products generate a high amount of value which is why they are referred to as the "80/20 rule" (Pareto principle). Whale customers, bulk orders, or viral products result in a long right tail.
Growth processes in the form of multiplication rather than addition. Many business phenomena compound and do not grow linearly (revenue growth, customer lifetime value, and social shares). Multiplicative processes are more likely to generate a right-skewed or log-normal distribution rather than a symmetric one.

- Random Variable refers to the variables where the numerical value depends upon the outcome of a random phenomenon or process. It does not mean that they are "random" but instead, the values that they can take have a particular probability of occurring.

Can take only a finite or countably infinite number of different possible values; you can enumerate the possibilities: 0, 1, 2, 3, ...
Examples: number of customers who leave this month, number of defectives in a sample, number of hits to a web site.
Defined by a probability mass function (PMF), which specifies the probability of each of those particular values: P(X=3)=0.15

P(H) — prior: your initial belief about the situation before new information
P(E | H) — likelihood: the probability of new information under the assumption that the hypothesis is true
P(H | E) — posterior: your updated belief based on the new information
P(E) — overall probability of new information, regardless of the hypothesis
Example calculation: fraud detection
P(transaction is fraudulent) = 0.1% — base rate across all transactions
New information: a fraud-detection algorithm detects the transaction. The algorithm detects 95% of fraudulent transactions (P(detection | fraud) = 0.95), but flags 2% of non-fraudulent transactions (P(detection | not fraud) = 0.02)
Bayes’ theorem incorporates these into P(fraud | detection) — the probability of the transaction being fraudulent, given that it was detected.
